# FEATURE ENGINEERING 

## 1. FEATURE SCALING 

## WHAT IS FEATURE SCALING 
* FEATURE SCALING IS A TECHNIQUE TO STANDARDIZE THE INDEPENDENT FEATURES PRESENT IN THE DATA IN A FIXED RANGE 

## WHY DO WE NEED FEATURE SCALING 
### FEATURE SCALING STANDARDIZES THE MAGNITUDE OF DIFFERENT DATASET ATTRIBUTES SO THAT SCALE- SENSITIVE ALGORITHMS LIKE GRADIENT DESCENTS , KNN, OR SVM COMPUTE DISTANCES AND UPDATES UNIFORMLY, PREVENTING LARGE SCALED FEATURES FROM DOMINATING MODEL TRAINING OR DISTORTING OPTIMISATION PATHS 

## TYPES OF FEATURE SCALING 
* STANDARDIZATION ALSO KNOWN AS Z-SCORE NORMALIZATION 
* NORMALIZATION

# STANDARDIZATION

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("Social_Network_Ads[1].csv")
```

```python
df.head()
df.shape
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(df.drop('Purchased',axis=1),df['Purchased'],test_size=0.3,random_state=0)
X_train.shape , X_test.shape
```

* df.drop(is me jo column dalo ke usko hata ke baki sab column pass honge
* train_test_split(input feaure x,output target y,test_size,random state)

# STANDARD SCALAR 

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaler.fit(X_train)  
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
scaler.mean_
```

* X_train is a pandas dataframe
* but X_train_scaled is a numpy array to solve this thing we have done this 

```python
X_train_scaled = pd.DataFrame(X_train_scaled,columns=X_test.columns)
X_test_scaled = pd.DataFrame(X_test_scaled , columns=X_test.columns)
```

```python
X_train_scaled
```

```python
np.round(X_train.describe(),1)
```

```python
np.round(X_train_scaled.describe(),1)
```

# EFFECT OF SCALING 

```python
fig, (ax1 , ax2)= plt.subplots(ncols=2,figsize=(12,5))
ax1.scatter(X_train['Age'],X_train['EstimatedSalary'])
ax1.set_title("BEFORE SCALING'-")
ax2.scatter(X_train_scaled['Age'],X_train_scaled['EstimatedSalary'],color='red')
ax2.set_title("AFTER SCALING")
plt.show
```
