## NORMALIZATION 

## normalization is a techinique often applied as part of data preapration for machine learning .The goal of normalization is to change values of numeric columns in dataset to use a common scale , without distorting differences in the range of values or losing information 

* it is always said that we must remove units in numeric columns and bring them in one scale 

## min max scaling 
## mean normalization 
## max absolute 
## robust scaling 

# MINMax scaling - (intution)

* Xi' = Xi-Xmin/Xmax-Xmin
* iski range [0,1] hogi bcx Xi=Xmin then Xi'=0
* if Xi=Xmax then Xi'=0 so its value will remain in the region [0,1]
  


```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('wine_data.csv', usecols=[0,1,2])
df.columns=['class label','Alcohol','malic acid']
```

```python
df

```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test= train_test_split(df.drop('class label',axis=1),df['class label'],test_size=0.3,random_state=0)
```

```python
X_train.shape ,X_test.shape
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler=MinMaxScaler()
scaler.fit(X_train)
X_train_scaled=scaler.transform(X_train)
X_test_scaled=scaler.transform(X_test)  #sickit learn ki library use karte hai toh woh dataframe ko numpy arrays me change kar deta hai 
```

```python
X_train_scaled=pd.DataFrame(X_train_scaled,columns=X_train.columns)
X_test_scaled=pd.DataFrame(X_test_scaled,columns=X_test.columns)
```

np.round(X_train.describe(),1)

# MEAN NORMALIZATION 

* Xi'= xi-xmean/Xmax-Xmin
* data is mean centered
* range is from [-1 1]
* -ve value means less than mean
* positive value means greater than mean
* rarely used no class in sckit learn
  

# Max absolute scaling 

* xi' = xi/|xmax|
* class in sckit learn as named maxabsscaled
* kisi bhi data me agar zeros zyada hai tab ye method use hota hai 

# Robust scaling 

* xi' = xi - xmedian/ interquartilerange(75th percentile - 25 percentile}
*  it is robust to outlier

# Normalization vs standardization
