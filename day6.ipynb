# fetching data using web scraping 

```python
import pandas as pd 
import requests 
from bs4 import BeautifulSoup
```

```python
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/119.0.0.0 Safari/537.36"
}
url="https://www.ambitionbox.com/list-of-companies?campaign=desktop_nav"
webpage=requests.get(url,headers=headers)
```

```python
soup=BeautifulSoup(webpage.text,parser="html parser")
```

```python
companies=soup.find_all("div",class_="companyCardWrapper")
name=[]
ratings=[]
reviews=[]
for i in companies:
    name.append(i.find("h2").text.strip())
    ratings.append(i.find("div",style= "height:auto;").text.strip())
    reviews.append(i.find("span",class_="companyCardWrapper__ActionCount").text.strip())
    
```
