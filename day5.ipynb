```python
df2 = []  # Empty list

for i in range(1, 429):
    url = f'https://api.themoviedb.org/3/movie/popular?api_key=dfe41487e4db0aa33ab62fd775811933&page={i}'
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/115.0.0.0 Safari/537.36"
    }
    
    # Try-Except se Connection Error handle kar rahe hain
    success = False
    while not success:
        try:
            r = requests.get(url, headers=headers, timeout=10)
            data = r.json()
            
            if 'results' in data:
                df_page = pd.DataFrame(data['results'])[['id', 'title', 'overview', 'release_date', 'popularity']]
                df2.append(df_page)
            
            success = True
        except Exception as e:
            print(f"Page {i} par issue aaya, 2 seconds baad retry kar rahe hain...")
            time.sleep(2)
            
    time.sleep(0.2)  # Delay taaki server block na kare

# DEKHO: Yeh dono lines loop ke BAHAR hain (no indentation)
final_df = pd.concat(df2, ignore_index=True)
final_df
```

```python
final_df.to_csv('movies.csv')
```
