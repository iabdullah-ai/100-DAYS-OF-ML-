# Categorical data

```python

df['Survived'].value_counts().plot(kind="bar")
```

# 2. Numerical data 

## a.Histogram

```python
import matplotlib.pyplot as plt
plt.hist(df['Age'],bins=10)
```

## b.Distplot

```python
sns.displot(data=df,x='Age',kde=True,stat='density') #ye bata hai ki data me skewnwss hai ki nhi 
```

## c.Boxplot

* q1= 25percentile
* q2=75 percentile
* interqualtile range = IQR= q3-q1
* minimum=q1-1.5*IQR
* maximum=q3+1.5*IQR
* agar koi value minimum aur maximum ke bahar hai toh hum use outliers kehte hai and we can eliminate it
* box plot hume batata hai ki kitna data noisy hai aur kitna data outliers hai

```python
sns.boxplot(data=df,x='Age')
```
